# fraud-eval on Hugging Face — quickstart

v0.1.0 smoke test, 20 cases, single test split. This is NOT the 190-case benchmark paper version.

Dataset: https://huggingface.co/datasets/RishabhPande/fraud-eval
Code: https://github.com/RishabhPande/fraud-eval

```python
from datasets import load_dataset

ds = load_dataset("RishabhPande/fraud-eval", split="test")
print(ds)
print(ds[0])
```

```python
from collections import Counter

print(Counter(ds['pack']))
print('fraud:', sum(ds['label']), 'benign:', len(ds) - sum(ds['label']))
```

```python
# Score a perfect detector, as a sanity check (detection 1.0, FPR 0.0)
refs = ds['label']
preds = refs
tp = sum(1 for p, r in zip(preds, refs) if p and r)
fp = sum(1 for p, r in zip(preds, refs) if p and not r)
print('detection:', tp / sum(refs), 'FPR:', fp / (len(refs) - sum(refs)))
```
